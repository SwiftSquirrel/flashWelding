```python
import pandas as pd
```

```python
def func(file_path: str) -> dict:
    data = pd.read_csv(file_path)
    file_path = data['file_path'].values[0]
    cols = list(data.columns)[:-1]
    return {file_path: cols}
```

```python
# 假设 feature_analysis 是一个目录，包含多个文件
import os

# 初始化一个空字典
result_dict = {}

# 遍历文件夹中的所有文件
feature_analysis_dir = "/Users/dawn/Desktop/HJ/feature_analysis"  # 替换为实际路径
for file_name in os.listdir(feature_analysis_dir):
    file_path = os.path.join(feature_analysis_dir, file_name)
    
    # 确保只处理文件
    if os.path.isfile(file_path):
        try:
            # 调用 func 函数
            func_result = func(file_path)
            
            # 更新 result_dict
            result_dict.update(func_result)
        except Exception as e:
            print(f"处理文件 {file_name} 时出错: {e}")

# 打印最终结果
print(result_dict)
```
